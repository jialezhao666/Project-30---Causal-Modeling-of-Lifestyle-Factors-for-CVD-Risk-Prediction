```python
# ============================================================
# Notebook 3 — Cell 1: Environment check (rpy2 + R + grf)
# Run this FIRST. Do not proceed until all three pass.
# ============================================================
import subprocess, sys

print("=== 1. rpy2 installed? ===")
try:
    import rpy2
    print(f"  rpy2 version: {rpy2.__version__}")
    RPY2_OK = True
except ImportError:
    print("  rpy2 NOT installed  ->  pip install rpy2")
    RPY2_OK = False

print("\n=== 2. R reachable from rpy2? ===")
if RPY2_OK:
    try:
        import rpy2.robjects as ro
        rver = ro.r('R.version.string')[0]
        print(f"  {rver}")
        R_OK = True
    except Exception as e:
        print(f"  R not reachable: {e}")
        R_OK = False
else:
    R_OK = False

print("\n=== 3. grf package available in R? ===")
if R_OK:
    try:
        import rpy2.robjects.packages as rpackages
        grf = rpackages.importr('grf')
        # grab grf version
        ver = ro.r('as.character(packageVersion("grf"))')[0]
        print(f"  grf version: {ver}")
        print("  multi_arm_causal_forest available:",
              bool(ro.r('exists("multi_arm_causal_forest")')[0]))
        GRF_OK = True
    except Exception as e:
        print(f"  grf NOT available: {e}")
        print('  In R, run:  install.packages("grf")')
        GRF_OK = False
else:
    GRF_OK = False

print("\n" + "="*50)
print(f"rpy2 : {'OK' if RPY2_OK else 'MISSING'}")
print(f"R    : {'OK' if R_OK else 'MISSING'}")
print(f"grf  : {'OK' if GRF_OK else 'MISSING'}")
print("="*50)
if RPY2_OK and R_OK and GRF_OK:
    print("All good. Proceed to Cell 2.")
else:
    print("Fix the missing piece before proceeding.")
```
